```python

import pandas as pd
import numpy as np

np.random.seed(42)
n = 250

distance = np.random.randint(20, 1200, n)
weight = np.round(np.random.uniform(1, 1000, n), 2)

traffic = np.random.choice(
    ["Low", "Medium", "High"], n, p=[0.3, 0.45, 0.25]
)
weather = np.random.choice(
    ["Clear", "Rainy", "Foggy"], n, p=[0.65, 0.25, 0.10]
)
vehicle = np.random.choice(
    ["Van", "Truck", "Bike"], n, p=[0.45, 0.35, 0.20]
)

speed = np.where(vehicle == "Bike", 35,
        np.where(vehicle == "Van", 48, 55))

traffic_factor = np.select(
    [traffic == "Low", traffic == "Medium", traffic == "High"],
    [1.0, 1.25, 1.6]
)
weather_factor = np.select(
    [weather == "Clear", weather == "Rainy", weather == "Foggy"],
    [1.0, 1.15, 1.3]
)

hours = np.maximum(
    distance / speed * traffic_factor * weather_factor
    + np.random.normal(0, 0.8, n),
    0.5
)

status = np.where(
    hours > distance / speed * 1.2, "Delayed", "On Time"
)

df = pd.DataFrame({
    "Shipment_ID": [f"SH{i:04d}" for i in range(1, n + 1)],
    "Distance_km": distance,
    "Weight_kg": weight,
    "Traffic_Level": traffic,
    "Weather_Condition": weather,
    "Vehicle_Type": vehicle,
    "Delivery_Time_Hours": np.round(hours, 2),
    "Delivery_Status": status
})

df.to_csv("week3_logistics_hypothetical_dataset.csv", index=False)

print("CSV created successfully!")
print("Total records:", len(df))
df.head()
```

```python
import pandas as pd

df = pd.read_csv("week3_logistics_hypothetical_dataset.csv")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
display(df.head())
```

```python

print("Missing values in each column:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python

print("1. Dataset Summary")
display(df.describe())

print("\n2. Delivery Status Count")
print(df["Delivery_Status"].value_counts())

print("\n3. Average Delivery Time")
print(round(df["Delivery_Time_Hours"].mean(), 2), "hours")

print("\n4. Average Distance")
print(round(df["Distance_km"].mean(), 2), "km")

print("\n5. Delivery Time by Traffic Level")
display(
    df.groupby("Traffic_Level")["Delivery_Time_Hours"]
      .mean()
      .round(2)
)

print("\n6. Delivery Time by Vehicle Type")
display(
    df.groupby("Vehicle_Type")["Delivery_Time_Hours"]
      .mean()
      .round(2)
)
```

```python

summary = pd.DataFrame({
    "Metric": [
        "Total Shipments",
        "Average Delivery Time (Hours)",
        "Average Distance (km)",
        "Delayed Shipments",
        "On-Time Shipments"
    ],
    "Value": [
        len(df),
        round(df["Delivery_Time_Hours"].mean(), 2),
        round(df["Distance_km"].mean(), 2),
        int((df["Delivery_Status"] == "Delayed").sum()),
        int((df["Delivery_Status"] == "On Time").sum())
    ]
})

summary.to_csv("week3_logistics_analysis_summary.csv", index=False)
print("Analysis summary saved successfully!")
display(summary)
```
